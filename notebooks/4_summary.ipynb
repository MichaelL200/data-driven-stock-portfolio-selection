# Podsumowanie

## Obserwacje, ograniczenia i wnioski

Projekt składał się z **dwóch głównych etapów** – utworzenia zbioru danych oraz opracowania i przetestowania strategii inwestycyjnych. Pierwszy z nich okazał się znacznie bardziej czasochłonny i złożony, ale jednocześnie kluczowy dla całego przedsięwzięcia, ponieważ bez odpowiedniej jakości danych wyniki strategii byłyby mało reprezentatywne. **Trudność** tego etapu wynikała z faktu, że żadne z wykorzystanych źródeł nie zapewniało pełnego pokrycia danych, a w przypadku serwisu EODHD zwiększanie pokrycia dla brakujących tickerów wymagało analizowania historii spółek i ręcznego mapowania ich oznaczeń.

Pomimo że celem projektu nie było pokonanie benchmarku (Sharpe ≈ 0,56), udało się to osiągnąć w przypadku dwóch utworzonych strategii:

1. Strategii najwyższych współczynników Sharpe’a przy doborze rocznym (Sharpe ≈ 0,65),

2. Strategii Monte Carlo na portfelach przy doborze kwartalnym (Sharpe ≈ 0,63).

Rezultaty te nie gwarantują jednak, że strategie przynosiłyby lepsze wyniki niż inwestowanie w szeroki indeks akcji, np. indeks S&P 500, z dwóch powodów:

1. Najbardziej fundamentalny z nich dotyczy **jakości danych**. Dane pobrane z obu serwisów nie mają pełnego pokrycia, co wprowadza błąd wyprzedzania, a próby jego ograniczenia również obarczone są niepewnością. Dodatkowo w utworzonym zbiorze danych mogą występować błędy wynikające ze sklejenia szeregów czasowych z dwóch źródeł, nieuwzględnienia splitów i scaleń akcji lub niejednolitego pokrycia danych w czasie. W przypadku nietypowych, skrajnych dziennych zmian cen, będących oczywistymi błędami danych, ich wpływ nie był uwzględniany podczas testowania stóp zwrotu portfeli.

2. Fakt, że kwartalna **strategia losowa** dawała lepsze zwroty niż benchmark jest sygnałem ostrzegawczym.

2. Nawet przy założeniu braku wpływu dwóvch powyższych czynników, historyczne wyniki strategii nie gwarantują podobnych rezultatów w przyszłości. **Przeszłe stopy zwrotu nie stanowią gwarancji przyszłych wyników**, a rynek finansowy podlega ciągłym zmianom.

## Przemyślenia i rozszerzenia

Sam projekt docelowo miał być bardziej rozbudowany i skoncentrowany na budowie strategii inwestycyjnych niż na przygotowaniu zbioru danych, jednak w praktyce ten drugi element okazał się niezbędny dla poprawnego działania całości. Z perspektywy czasu uważam, że w tego typu projektach warto jak najszybciej zbudować minimalną, działającą wersję systemu, nawet jeśli początkowo jest ona mocno uproszczona.

Moim wnioskiem z pracy nad projektem jest to, że bardziej efektywne jest **podejście iteracyjne** – rozpoczęcie od najprostszej wersji pipeline’u danych i strategii, a następnie ich stopniowe rozszerzanie. Pozwala to szybciej uzyskiwać pierwsze rezultaty, lepiej rozumieć ograniczenia systemu oraz sprawniej identyfikować kluczowe problemy.

W obecnym podejściu prace były realizowane w **sposób sekwencyjny**, z naciskiem na maksymalne dopracowanie kolejnych etapów przed przejściem do następnych. Z perspektywy projektu okazało się to mniej elastyczne i wydłużyło moment uzyskania działających rezultatów. W przyszłości bardziej naturalne wydaje się rozpoczęcie od prostego rdzenia systemu, a następnie jego stopniowe rozwijanie w kierunku elementów najbardziej wpływających na jakość wyników.

Istniejący projekt stanowi solidną bazę, którą można w przyszłości **rozszerzyć** o bardziej zaawansowane podejścia do budowy strategii inwestycyjnych. Naturalnym kierunkiem dalszego rozwoju jest uwzględnienie metod opartych na analizie technicznej oraz fundamentalnej, a także wykorzystanie bardziej złożonych technik uczenia maszynowego do selekcji spółek i optymalizacji portfela.
